```python
# Core imports and plotting backend setup
import os
import pandas as pd
import rhkpy
import holoviews as hv
import requests

hv.extension('bokeh')
```

```python
# Download source files from Zenodo if missing
ZENODO_FILES = {
    'didv_clean.sm4': 'https://zenodo.org/records/17469441/files/dI-dV_hopg_9K_2021_04_13_15_14_15_591.sm4?download=1',
    'didv_alkane.sm4': 'https://zenodo.org/records/17469441/files/KK_MK_FLG_ABC_73_9K_2024_06_27_21_05_08_634.sm4?download=1',
}

for local, url in ZENODO_FILES.items():
    if url is None or os.path.exists(local):
        continue
    response = requests.get(url, timeout=30)
    response.raise_for_status()
    with open(local, 'wb') as fh:
        fh.write(response.content)
```

```python
# Helper utilities to prepare spectra DataFrames
def _mean_lia(dataset, *, dist_slice=None):
    lia = dataset.spectra.variables['lia']
    if dist_slice is not None and 'dist' in lia.dims:
        lia = lia.isel(dist=slice(*dist_slice))
    dims_to_average = [dim for dim in lia.dims if dim != 'bias']
    if dims_to_average:
        lia = lia.mean(dim=dims_to_average)
    return lia.to_numpy()


def load_clean_spectrum(path):
    dataset = rhkpy.rhkdata(path)
    bias = dataset.spectra.coords['bias'].to_numpy()
    lia = _mean_lia(dataset)
    span = lia.max() - lia.min() if lia.size else 0.0
    if span:
        lia = (lia - lia.min()) / span
    return pd.DataFrame({'bias': bias, 'lia': lia})


def load_contaminated_spectrum(path):
    dataset = rhkpy.rhkdata(path)
    bias = dataset.spectra.coords['bias'].to_numpy() + 0.02
    lia = _mean_lia(dataset, dist_slice=(60, 80))
    if lia.ndim > 1:
        lia = lia.mean(axis=tuple(range(1, lia.ndim)))
    scale = lia.max() * 1.5 if lia.size else 1.0
    if not scale:
        scale = 1.0
    lia = lia / scale + 0.1
    return pd.DataFrame({'bias': bias, 'lia': lia})
```

```python
# Compose final Holoviews overlay with the Bokeh backend
clean_df = load_clean_spectrum('didv_clean.sm4')
contaminated_df = load_contaminated_spectrum('didv_alkane.sm4')

def dashed_segment(x0, y0, y1, color, label=None):
    segment = hv.Segments([(x0, y0, x0, y1)]).opts(color=color, line_dash='dashed', line_width=1.2)
    return segment.relabel(label) if label else segment

clean_curve = hv.Curve(clean_df, 'bias', 'lia', label='UHV cleaved').opts(color='blue', line_width=1.6)
contaminated_curve = hv.Curve(contaminated_df, 'bias', 'lia', label='Contaminated').opts(color='red', line_width=1.6)

energy_guides = hv.Overlay([
    dashed_segment(-0.0568, 0, 0.4, 'darkgreen'),
    dashed_segment(0.0568, 0, 0.4, 'darkgreen', '56.8 meV'),
    dashed_segment(-0.0809, 0, 0.4, 'lime'),
    dashed_segment(0.0809, 0, 0.4, 'lime', '80.9 meV'),
]).opts(show_legend=True)

plot_spectra = (clean_curve * contaminated_curve * energy_guides).opts(
    xlabel='sample bias (V)',
    ylabel='dI/dV signal (a.u)',
    xlim=(-0.5, 0.5),
    ylim=(-0.05, 0.95),
    width=550,
    height=360,
    legend_position='top_left',
    fontscale=0.9,
)
plot_spectra
```
