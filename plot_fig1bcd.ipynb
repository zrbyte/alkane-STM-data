# STM Topography Images
Comparing topography images for clean, contaminated, and alkane-covered graphite

```python
import os
import numpy as np
import xarray as xr
import holoviews as hv
import rhkpy
import requests

hv.extension('bokeh')
```

```python
import sys
print(sys.version)
print(f'numpy version: {np.__version__}')
print(f'xarray version: {xr.__version__}')
print(f'holoviews version: {hv.__version__}')
print(f'rhkpy version: {rhkpy.__version__}')
```

```python
# Download source files from Zenodo if missing
ZENODO_FILES = {
    'stm_clean.gwy.txt': 'https://zenodo.org/records/17469441/files/MK_ABC_FLG_25_9K_2023_08_01_12_52_49_951.sm4_2.gwy.txt?download=1',
    'stm_cont.gwy.txt': 'https://zenodo.org/records/17469441/files/Kalvin_HOPG_UHV_exf240717_9K_2024_07_17_14_20_32_911.sm4.gwy.txt?download=1',
    'stm_alk.gwy.txt': 'https://zenodo.org/records/17469441/files/Stripes-9K-HOPG-SPI2-3_2021_09_07_00_36_20_114.sm4.gwy.txt?download=1',
}

for local, url in ZENODO_FILES.items():
    if url is None or os.path.exists(local):
        continue
    print(f'Downloading {local}...')
    response = requests.get(url, timeout=60)
    response.raise_for_status()
    with open(local, 'wb') as fh:
        fh.write(response.content)
    print(f'✓ {local} downloaded')
```

```python
# Load topography data (exported from Gwyddion as .txt)
gr_clean = np.loadtxt('stm_clean.gwy.txt') * 1e12
gr_cont = np.loadtxt('stm_cont.gwy.txt') * 1e12
gr_alk = np.loadtxt('stm_alk.gwy.txt') * 1e12

# Create coordinate arrays
x1, y1 = np.linspace(0, 5, 512), np.linspace(0, 5, 512)
x2, y2 = np.linspace(0, 5, 512), np.linspace(0, 5, 512)
x3, y3 = np.linspace(0, 25, 1024), np.linspace(0, 25, 1024)
```

```python
# Clean graphite image
img = hv.Image((x1, y1, gr_clean)).opts(
    clim=(5, 40),
    xlabel='',
    ylabel='',
    colorbar=True,
    cmap='viridis',
    tools=['hover'],
    title='Clean Graphite',
    aspect=1
)

img
```

```python
# Contaminated graphite image
img2 = hv.Image((x2, y2, gr_cont)).opts(
    clim=(5, 15),
    xlabel='',
    ylabel='',
    colorbar=True,
    cmap='viridis',
    tools=['hover'],
    title='Contaminated Graphite',
    aspect=1
)

img2
```

```python
# Alkane-covered graphite image
img3 = hv.Image((x3, y3, gr_alk)).opts(
    clim=(100, 250),
    xlabel='',
    ylabel='',
    colorbar=True,
    cmap='viridis',
    tools=['hover'],
    title='Alkane-Covered Graphite',
    aspect=1
)

img3
```
